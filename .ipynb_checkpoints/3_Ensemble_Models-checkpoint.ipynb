```python
import pandas as pd
import joblib
import os
from imblearn.over_sampling import SMOTE

# 1. Nạp dữ liệu sạch
df_train = pd.read_csv('data/processed/train_final.csv').dropna()
df_val = pd.read_csv('data/processed/val_final.csv').dropna()

# 2. Tải Vectorizer cũ (bắt buộc để đồng nhất từ điển)
tfidf = joblib.load('saved_models/tfidf_vectorizer.pkl')

X_train = tfidf.transform(df_train['text_cleaned'])
X_val = tfidf.transform(df_val['text_cleaned'])

y_train = df_train['label']
y_val = df_val['label']

# 3. SMOTE cân bằng dữ liệu 1:1 trên tập Train
smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print(f"✅ Đã chuẩn bị dữ liệu. Kích thước tập Train sau SMOTE: {X_train_res.shape}")
print(f"Phân bổ nhãn: \n{y_train_res.value_counts()}")
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

print("--- 1. TINH CHỈNH LOGISTIC REGRESSION ---")

param_grid_lr = {
    'C': [0.1, 1.0, 10.0],
    'class_weight': ['balanced']
}
# Tổng cộng: 5 ứng viên x 3-fold = 15 lần chạy

grid_lr = GridSearchCV(LogisticRegression(max_iter=1000), param_grid_lr, cv=3, scoring='f1', verbose=1)
grid_lr.fit(X_train_res, y_train_res)

best_lr = grid_lr.best_estimator_
print(f"🔥 Tham số tốt nhất LR: {grid_lr.best_params_}")
print(f"F1-Score trên Val: {grid_lr.best_score_:.4f}")
```

```python
from sklearn.ensemble import RandomForestClassifier

print("\n--- 2. TINH CHỈNH RANDOM FOREST ---")

param_grid_rf = {
    'max_depth': [10, 20, 30, None], # Khống chế độ sâu cây
    'n_estimators': [100, 200],
    'min_samples_split': [2, 5, 10],
    'class_weight': ['balanced']
}

grid_rf = GridSearchCV(RandomForestClassifier(random_state=42, n_jobs=-1), param_grid_rf, cv=3, scoring='f1', verbose=1)
grid_rf.fit(X_train_res, y_train_res)

best_rf = grid_rf.best_estimator_
print(f"🔥 Tham số tốt nhất RF: {grid_rf.best_params_}")
print(f"F1-Score trên Val: {grid_rf.best_score_:.4f}")
```

```python
from xgboost import XGBClassifier

print("\n--- 3. TINH CHỈNH XGBOOST ---")

param_grid_xgb = {
    'max_depth': [3, 6, 10],
    'learning_rate': [0.01, 0.1, 0.2],
    'n_estimators': [100, 200]
}
# Tổng cộng: 162 ứng viên x 3-fold = 486 lần chạy (Sẽ chạy hơi lâu một chút)
grid_xgb = GridSearchCV(XGBClassifier(random_state=42), param_grid_xgb, cv=3, scoring='f1', verbose=1)
grid_xgb.fit(X_train_res, y_train_res)

best_xgb = grid_xgb.best_estimator_
print(f"🔥 Tham số tốt nhất XGB: {grid_xgb.best_params_}")
print(f"F1-Score trên Val: {grid_xgb.best_score_:.4f}")
```

```python
# Lưu lại 3 model đã tối ưu
joblib.dump(best_lr, 'saved_models/best_lr.pkl')
joblib.dump(best_rf, 'saved_models/best_rf.pkl')
joblib.dump(best_xgb, 'saved_models/best_xgb.pkl')

print("🚀 ĐÃ LƯU 3 MÔ HÌNH TỐI ƯU SẴN SÀNG CHO FILE 4!")
```
