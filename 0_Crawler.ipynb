```python
from youtube_comment_downloader import YoutubeCommentDownloader, SORT_BY_RECENT
import pandas as pd
import time

# 1. Danh sách các link bạn đã chọn
video_urls = [
    'https://youtu.be/K9RFqA4bNfg?si=siFl_m8dpY_mLZ8C', # Video 1
    'https://youtu.be/lXYHUsiKqcU?si=JfnWnaDR8XwBPh-u', # Video 2
    'https://www.youtube.com/watch?v=z7wKP-3Mabw'  # Video 3
]

downloader = YoutubeCommentDownloader()
all_comments = []

print("--- BẮT ĐẦU CÀO DỮ LIỆU TỪ 3 VIDEO ---")

for url in video_urls:
    print(f"Đang cào video: {url}...")
    try:
        # Lấy comment từ mỗi video
        generator = downloader.get_comments_from_url(url, sort_by=SORT_BY_RECENT)
        
        count = 0
        for comment in generator:
            all_comments.append({
                'comment': comment['text'],
                #'video_url': url # Lưu lại nguồn để sau này kiểm chứng
            })
            count += 1
            
            # Mỗi video lấy khoảng 300-400 câu để tổng đạt ~1000 câu
            if count >= 500: 
                break
        
        print(f"✅ Đã lấy được {count} comment.")
        time.sleep(2) # Nghỉ 2 giây để tránh bị YouTube chặn (Rate Limit)
        
    except Exception as e:
        print(f"❌ Lỗi khi cào video {url}: {e}")

# 2. Chuyển thành DataFrame
df_yt = pd.DataFrame(all_comments)

# 3. Thêm cột label trống để bạn ngồi điền 0 hoặc 1
df_yt['label'] = ""

# 4. Lưu ra file CSV (Dùng utf-8-sig để mở bằng Excel không bị lỗi font tiếng Việt)
output_file = 'data/youtube_raw.csv'
df_yt.to_csv(output_file, index=False, encoding='utf-8-sig')

print(f"\n🚀 HOÀN TẤT! Tổng cộng cào được {len(df_yt)} dòng.")
print(f"File lưu tại: {output_file}")
print("Bây giờ bạn hãy mở file này bằng Excel và bắt đầu gán nhãn vào cột 'label' nhé!")
```
